Problem1: A fair coin is tossed 100 times.

Calculate:

Number of heads

 Number of tails

 Experimental probability of Head

 Experimental probability of Tail

```python
import random

def coin_toss(n):
  heads = 0
  tails = 0

  for i in range(n):
    result = random.randint(0,1)

    if result == 1:
      heads += 1
    else:
      tails += 1

  print("\nNumber of tosses: ", n)
  print("Number of heads: ", heads)
  print("Number of tails: ", tails)
  print("Experimental probability of Head : ", round(heads/n, 4))
  print("Experimental probability of Tail : ", round(tails/n, 4))

coin_toss(100)
```

Problem 2: check Effect of the number of trials
Repeat the coin-toss experiment for:

n=10, 100, 1,000, 10,000

```python
coin_toss(10)
coin_toss(100)
coin_toss(1000)
coin_toss(10000)
```

Rolling a die: A fair die is rolled 6,000 times.

Calculate the experimental probability of each outcome: S={1,2,3,4,5,6}

```python
import random

n = 6000
counts = [0] * 6

for i in range(n):
  result = random.randint(1,6)
  counts[result-1] += 1

for i in range(6):
  print("\n Outcome: ", i+1)
  print("counts: ", counts[i])
  print("experimental probability: ",counts[i]/n)
```

Addition rule for disjoint events

 A die is rolled 10,000 times.

 Define: A=getting 1, B= getting 6

 check the events are joint or disjoint

Verify: P(A∪B)=P(A)+P(B)

```python
import random

n = 10000
count_A = 0
count_B = 0
count_A_and_B = 0
count_union = 0

for i in range(n):
    result = random.randint(1, 6)

    if result == 1:
        count_A += 1

    if result == 6:
        count_B += 1

    # Check whether both events occur together
    if result == 1 and result == 6:
        count_A_and_B += 1

    if result == 1 or result == 6:
        count_union += 1

P_A = count_A / n
P_B = count_B / n
P_intersection = count_A_and_B / n
P_union = count_union / n

# Check whether events are disjoint or joint
if count_A_and_B == 0:
    print("Events A and B are DISJOINT")
else:
    print("Events A and B are JOINT")

# Check the formula: P(A or B) = P(A) + P(B) - P(A and B)
RHS = P_A + P_B - P_intersection

print("P(A):", P_A)
print("P(B):", P_B)
print("P(A intersection B):", P_intersection)

print("To verify: P(A_union_B) = P_A + P_B ")
print("P(A union B):", P_union)
print(f"P(A) + P(B):, {P_A + P_B: .4f}")

print(f"Verified?  : {P_union == RHS}")
```

Addition rule for overlapping events

For one die roll, define: A=getting an even number A={2,4,6} and B=getting a number greater than 3 B={4,5,6}

Verify: P(A∪B)=P(A)+P(B)−P(A∩B)

```python
S = {1, 2, 3, 4, 5, 6}

# Define events
A = {2, 4, 6}       # Getting an even number
B = {4, 5, 6}       # Getting a number greater than 3

# Find union and intersection
A_union_B = A | B
A_intersection_B = A & B

# Calculate probabilities
P_A = len(A) / len(S)
P_B = len(B) / len(S)
P_intersection = len(A_intersection_B) / len(S)
P_union = len(A_union_B) / len(S)

# Verify the addition rule
RHS = P_A + P_B - P_intersection

print("A:", A)
print("B:", B)
print("Events are OVERLAPPING")

print(f"P(A): {P_A:.4f}")
print(f"P(B): {P_B:.4f}")
print(f"P(A intersection B): {P_intersection:.4f}")
print(f"P(A union B): {P_union:.4f}")
print(f"P(A) + P(B) - P(A intersection B): {RHS:.4f}")

print(f"Verified?   : {round(P_union, 4) == round(RHS, 4)}")
```

Multiplication rule for independent events

A coin is tossed and a die is rolled 10,000 times.
 Define:
 obtaining Head
 obtaining an even number

 Verify: P(A∩B)=P(A)P(B)

```python
import random

n = 10000
count_A = 0
count_B = 0
count_A_and_B = 0

# Toss a coin and roll a die 10,000 times
for i in range(n):
    coin = random.choice(["H", "T"])
    die = random.randint(1, 6)

    # Event A: Obtaining Head
    if coin == "H":
        count_A += 1

    # Event B: Obtaining an even number
    if die in {2, 4, 6}:
        count_B += 1

    # Both A and B occur
    if coin == "H" and die in {2, 4, 6}:
        count_A_and_B += 1

# Calculate experimental probabilities
P_A = count_A / n
P_B = count_B / n
P_intersection = count_A_and_B / n

# Verify multiplication rule
RHS = P_A * P_B

print(f"P(A): {P_A:.4f}")
print(f"P(B): {P_B:.4f}")
print(f"P(A intersection B): {P_intersection:.4f}")
print(f"P(A) * P(B): {RHS:.4f}")

print(f"Verified ?   : {round(P_intersection, 4) == round(RHS,4)}")
```
