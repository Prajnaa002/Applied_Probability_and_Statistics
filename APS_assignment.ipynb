```python
import random

def coin_toss(n):
    heads = 0
    tails = 0

    for x in range(n):
        result = random.randint(0, 1)

        if result == 1:
            heads += 1
        else:
            tails += 1

    print("Total number of tosses:", n)
    print("Heads:", heads)
    print("Tails:", tails)

    print("Experimental Probability of Heads:", round(heads / n, 2))
    print("Experimental Probability of Tails:", round(tails / n, 2))

coin_toss(100)
```

```python
#2
coin_toss(10)
coin_toss(100)
coin_toss(1000)
coin_toss(10000)
```

```python
import random

# Roll the die 10000 times
total_rolls = 10000

rolls = [random.randint(1, 6) for x in range(total_rolls)]

# Count how many times we get 1 and 6
count_A = rolls.count(1)  # Getting 1
count_B = rolls.count(6)  # Getting 6

# We cannot get 1 and 6 in one roll
count_A_and_B = sum( 1 for r in rolls if r == 1 and r == 6)

# Count when the result is either 1 or 6
count_A_or_B = sum( 1 for r in rolls if r == 1 or r == 6)

# Calculate the probabilities
P_A = count_A / total_rolls
P_B = count_B / total_rolls
P_A_and_B = count_A_and_B / total_rolls
P_A_or_B = count_A_or_B / total_rolls

# Check the formula: P(A or B) = P(A) + P(B) - P(A and B)
RHS = P_A + P_B - P_A_and_B

print(f"Observed P(A)               : {P_A:.4f}")
print(f"Observed P(B)               : {P_B:.4f}")
print(f"Observed P(A ∩ B)           : {P_A_and_B:.4f}")
print(f"Observed P(A ∪ B)           : {P_A_or_B:.4f}")
print(f"Calculated P(A)+P(B)-P(A∩B) : {RHS:.4f}")
print(f"Verified?                   : {P_A_or_B == RHS}")
```

```python
import random

total_rolls = 10000

# Roll a dice 10000 times
rolls = [random.randint(1, 6) for x in range(total_rolls)]

# A: getting an even number
count_A = sum(1 for r in rolls if r in {2, 4, 6})
# B: getting a number greater than 3
count_B = sum(1 for r in rolls if r in {4, 5, 6})

# A and B: numbers that are both even and greater than 3
count_A_and_B = sum(1 for r in rolls if (r in {2, 4, 6}) and (r in {4, 5, 6}))

# A or B: numbers that are either even or greater than 3
count_A_or_B = sum(1 for r in rolls if (r in {2, 4, 6}) or (r in {4, 5, 6}))

# Find the experimental probabilities
P_A = count_A / total_rolls
P_B = count_B / total_rolls
P_A_and_B = count_A_and_B / total_rolls
P_A_or_B = count_A_or_B / total_rolls

RHS = P_A + P_B - P_A_and_B

print(f"Observed P(A)               : {P_A:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(B)               : {P_B:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(A ∩ B)           : {P_A_and_B:.4f}  (Theoretical: 0.3333)")
print(f"Observed P(A ∪ B)           : {P_A_or_B:.4f}  (Theoretical: 0.6667)")
print(f"Calculated P(A)+P(B)-P(A∩B) : {RHS:.4f}")
print(f"Verified?                   : {round(P_A_or_B, 4) == round(RHS, 4)}")
```

```python
import random

total_trials = 10000

# Do 10000 coin flips and dice rolls

trials = [(random.randint(0, 1), random.randint(1, 6)) for x in range(total_trials)]

# Count heads
count_A = sum(1 for coin, die in trials if coin == 1)

# Count even numbers on the dice
count_B = sum(1 for coin, die in trials if die % 2 == 0)

# Count cases where we get both head and even
count_A_and_B = sum(1 for coin, die in trials if coin == 1 and die % 2 == 0)

# Calculate probabilities
P_A = count_A / total_trials
P_B = count_B / total_trials
P_A_and_B = count_A_and_B / total_trials

# Multiplication Rule RHS calculation
RHS = P_A * P_B

print(f"Observed P(A)           : {P_A:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(B)           : {P_B:.4f}  (Theoretical: 0.5000)")
print(f"Observed P(A ∩ B)       : {P_A_and_B:.4f}  (Theoretical: 0.2500)")
print(f"Calculated P(A) * P(B)  : {RHS:.4f}  (Theoretical: 0.2500)")
print(f"Verified?               : {round(P_A_and_B, 3) == round(RHS, 3)}")
```
