#coding exercise: Measure of Central Tendency



```python
import random
from collections import Counter
import numpy as np
from scipy import stats # used for comparison of mode

# 1. Generate 10000 random samples (integers between 1 and 100)
random.seed(42)  # Setting seed for reproducibility
samples = [random.randint(1, 100) for _ in range(10000)]

# 2. Compute mean, median, mode WITHOUT using numpy
print("=== Without NumPy ===")
# Mean
mean_pure = sum(samples) / len(samples)
print(f"Mean: {mean_pure}")

# Median
sorted_samples = sorted(samples)
n = len(sorted_samples)
if n % 2 == 1:
    median_pure = sorted_samples[n // 2]
else:
    median_pure = (sorted_samples[(n // 2) - 1] + sorted_samples[n // 2]) / 2.0
print(f"Median: {median_pure}")

# Mode
counts = Counter(samples)
max_count = max(counts.values())
mode_pure = [k for k, v in counts.items() if v == max_count]
print(f"Mode: {mode_pure} (Frequency: {max_count})")

# 3. Compute mean, median, mode USING numpy
print("\n=== With NumPy/SciPy ===")
mean_np = np.mean(samples)
median_np = np.median(samples)
# np.argmax(np.bincount()) or stats.mode can be used for mode
mode_np = stats.mode(samples, keepdims=True)

print(f"Mean: {mean_np}")
print(f"Median: {median_np}")
print(f"Mode: {mode_np.mode[0]} (Frequency: {mode_np.count[0]})")
```

#Coding exercise: Measure of Dispersion

```python
import random
import math
import numpy as np
from scipy import stats

# 1. Generate 10000 random samples (integers between 1 and 100)
random.seed(42)
samples = [random.randint(1, 100) for _ in range(10000)]

# 2. Compute Dispersion Metrics WITHOUT using external numpy/scipy libraries
print("=== Without NumPy/SciPy (Pure Python) ===")

# Range
rng_pure = max(samples) - min(samples)
print(f"Range: {rng_pure}")

# Variance (using population variance formula: sum((x - mean)^2) / N)
mean_val = sum(samples) / len(samples)
var_pure = sum((x - mean_val) ** 2 for x in samples) / len(samples)
print(f"Variance: {var_pure:.4f}")

# Standard Deviation
std_pure = math.sqrt(var_pure)
print(f"Standard Deviation: {std_pure:.4f}")

# IQR (Interquartile Range) using the standard percentile calculation
sorted_samples = sorted(samples)
def get_percentile(data, percentile):
    k = (len(data) - 1) * percentile
    f = math.floor(k)
    c = math.ceil(k)
    if f == c:
        return data[int(k)]
    return data[int(f)] * (c - k) + data[int(c)] * (k - f)

q1_pure = get_percentile(sorted_samples, 0.25)
q3_pure = get_percentile(sorted_samples, 0.75)
iqr_pure = q3_pure - q1_pure
print(f"IQR: {iqr_pure:.4f} (Q1: {q1_pure:.1f}, Q3: {q3_pure:.1f})")

# 3. Compute Dispersion Metrics USING NumPy and SciPy
print("\n=== With NumPy/SciPy ===")
rng_np = np.ptp(samples)
var_np = np.var(samples)  # population variance by default
std_np = np.std(samples)
iqr_sp = stats.iqr(samples)

print(f"Range: {rng_np}")
print(f"Variance: {var_np:.4f}")
print(f"Standard Deviation: {std_np:.4f}")
print(f"IQR: {iqr_sp:.4f}")
```
