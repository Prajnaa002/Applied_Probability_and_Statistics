#Generate 10000 random sample normally distributed. Calculate and plot the population mean, standard deviation. Draw one random sample of 30 observations. Calculate sample mean, sample standard deviation and standard error.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

population = np.random.normal(loc=50, scale=10, size=10000)

# population mean and standard deviation
pop_mean = np.mean(population)
pop_std = np.std(population)

print("Population Mean:", pop_mean)
print("Population Standard Deviation:", pop_std)

# Plot population distribution
plt.figure(figsize=(10, 5))

plt.hist(population, bins=50, density=True, alpha=0.6,
         edgecolor='black', label='Population Distribution')

# Plot normal distribution curve
x = np.linspace(min(population), max(population), 200)
plt.plot(x, norm.pdf(x, pop_mean, pop_std),
         linewidth=2, label='Normal Distribution Curve')

plt.axvline(pop_mean, linestyle='--',
            label='Population Mean')

plt.xlabel("Values")
plt.ylabel("Density")
plt.title("Population Distribution (10000 Observations)")
plt.legend()
plt.show()

#Draw one random sample of 30 observations
sample = np.random.choice(population, size=30, replace=False)

#sample statistics
sample_mean = np.mean(sample)
sample_std = np.std(sample, ddof=1)

# standard error
standard_error = sample_std / np.sqrt(len(sample))

print("\nSample Observations:\n", sample)
print("\nSample Mean:", sample_mean)
print("Sample Standard Deviation:", sample_std)
print("Standard Error:", standard_error)

#Plot
plt.figure(figsize=(10, 5))
plt.hist(sample, bins=10, alpha=0.7, edgecolor='black')

plt.axvline(sample_mean, color='red', linestyle='--',
            label='Sample Mean')

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Sample Distribution (30 Observations)")
plt.legend()
plt.show()
```
